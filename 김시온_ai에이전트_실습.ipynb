```python
from google import genai
from google.genai import types
from google.colab import userdata
key = userdata.get('API_KEY')

client = genai.Client(api_key=key)



# ① AI 응답을 받아오는 함수
def get_ai_response(history_list):
    response = client.models.generate_content(
        model="models/gemini-2.5-flash",
        contents=history_list,  # 누적된 대화 기록 리스트를 통째로 전달
        config=types.GenerateContentConfig(
            temperature=0.9,
            system_instruction="너는 우주의 모든 진실을 알고있는 전지한 신 비슷한거야."  # 시스템 프롬프트는 따로 분리
        )
    )
    return response.text

# Gemini는 시스템 프롬프트를 config에 따로 설정하므로, 대화 기록은 빈 리스트로 시작합니다.
messages = []

while True:
    user_input = input("사용자: ")

    if user_input.lower() == "exit":  # ② 종료 조건
        print("대화를 종료합니다.")
        break

    # 1. 사용자 메시지를 대화 기록 리스트에 추가 (역할: user)
    messages.append(
        types.Content(role="user", parts=[types.Part.from_text(text=user_input)])
    )

    # 2. 누적된 대화 기록을 보내서 AI 응답 받아오기
    ai_response = get_ai_response(messages)

    # 3. 받아온 AI 응답도 대화 기록 리스트에 추가 (역할: model)
    messages.append(
        types.Content(role="model", parts=[types.Part.from_text(text=ai_response)])
    )

    print("AI: " + ai_response)
```

```python
from google import genai
from google.genai import types
from google.colab import userdata

key = userdata.get('secretName')
client = genai.Client(api_key=key)

# ① AI 응답을 스트리밍으로 받아오면서 화면에 즉시 출력하는 함수
def get_stream_ai_response(history_list):
    # generate_content -> generate_content_stream 으로 변경
    response_stream = client.models.generate_content_stream(
        model="models/gemini-2.5-flash",
        contents=history_list,  # 누적된 대화 기록 리스트를 통째로 전달
        config=types.GenerateContentConfig(
            temperature=0.9,
            system_instruction="너는 사용자를 도와주는 상담사야."  # 시스템 프롬프트는 따로 분리
        )
    )

    full_text = ""
    print("AI: ", end="", flush=True)  # AI 응답 출력 시작 (줄바꿈 없이)

    for chunk in response_stream:
        # 텍스트 조각(chunk)이 도착할 때마다 화면에 즉시 출력
        print(chunk.text, end="", flush=True)
        full_text += chunk.text  # 대화 기록에 저장하기 위해 전체 텍스트 누적

    print()  # 응답이 모두 끝나면 줄바꿈 처리

    return full_text  # 누적된 전체 텍스트 반환

# Gemini는 시스템 프롬프트를 config에 따로 설정하므로, 대화 기록은 빈 리스트로 시작합니다.
messages = []

while True:
    user_input = input("사용자: ")

    if user_input.lower() == "exit":  # ② 종료 조건
        print("대화를 종료합니다.")
        break

    # 1. 사용자 메시지를 대화 기록 리스트에 추가 (역할: user)
    messages.append(
        types.Content(role="user", parts=[types.Part.from_text(text=user_input)])
    )

    # 2. 누적된 대화 기록을 보내서 AI 응답을 스트리밍으로 출력하고 전체 문장 반환받기
    # (함수 내부에서 화면 출력이 이루어지므로 메인 루프에서는 따로 print하지 않습니다)
    ai_response = get_stream_ai_response(messages)

    # 3. 받아온 AI 응답(전체 문장)도 대화 기록 리스트에 추가 (역할: model)
    messages.append(
        types.Content(role="model", parts=[types.Part.from_text(text=ai_response)])
    )
```

```python
%%writefile app.py
import streamlit as st
from google import genai
from google.genai import types

# ❌ 삭제된 부분: from google.colab import userdata

# ✅ 수정된 부분: Streamlit secrets에서 API 키 불러오기
api_key = st.secrets["GEMINI_API_KEY"]
client = genai.Client(api_key=api_key)

st.title("")
st.write(".")

def stream_ai_response(messages_dict):
    history_list = []
    for msg in messages_dict:
        role = "model" if msg["role"] == "assistant" else "user"
        history_list.append(
            types.Content(role=role, parts=[types.Part.from_text(text=msg["content"])])
        )

    response_stream = client.models.generate_content_stream(
        model="models/gemini-2.5-flash",
        contents=history_list,
        config=types.GenerateContentConfig(
            temperature=0.9,
            system_instruction="너는 ."
        )
    )

    for chunk in response_stream:
        yield chunk.text

if "messages" not in st.session_state:
    st.session_state.messages = []

for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

if prompt := st.chat_input("메시지를 입력하세요..."):
    st.session_state.messages.append({"role": "user", "content": prompt})

    with st.chat_message("user"):
        st.markdown(prompt)

    with st.chat_message("assistant"):
        response_generator = stream_ai_response(st.session_state.messages)
        ai_response = st.write_stream(response_generator)

    st.session_state.messages.append({"role": "assistant", "content": ai_response})
```

```python
import pymupdf
import os

pdf_file_path = "/content/암흑물질 탐색을 위한 저온에서 CsI 결정의 특성 연구 (1).txt" # 경로 수정
doc = pymupdf.open(pdf_file_path)

header_height = 80
footer_height = 80

full_text = ''

for page in doc:
    rect = page.rect # 페이지 크기 가져오기

    header = page.get_text(clip=(0, 0, rect.width , header_height))
    footer = page.get_text(clip=(0, rect.height - footer_height, rect.width , rect.height))
    text = page.get_text(clip=(0, header_height, rect.width , rect.height - footer_height))

    full_text += text + '\n------------------------------------\n'

# 파일명만 추출
pdf_file_name = os.path.basename(pdf_file_path)
pdf_file_name = os.path.splitext(pdf_file_name)[0] # 확장자 제거

txt_file_path = f'{pdf_file_name}_with_preprocessing.txt'

with open(txt_file_path, 'w', encoding='utf-8') as f:
    f.write(full_text)
```

```python
from google import genai
from google.genai import types
from google.colab import userdata # 코랩 보안 키 사용

def summarize_txt(file_path: str):
    # 클라이언트 생성 (코랩 비밀번호에 저장된 키 사용)
    api_key = userdata.get('API_KEY')
    client = genai.Client(api_key=api_key)

    if not os.path.exists(file_path):
        return "파일을 찾을 수 없습니다. 경로를 확인해주세요."

    with open(file_path, 'r', encoding='utf-8') as f:
        txt = f.read()

    system_instruction = '''
    너는 문서를 요약하는 봇이다. 사용자가 제공하는 글을 읽고, 저자의 문제 인식과 주장을 파악하여 주요 내용을 요약하라.

    작성해야 하는 포맷은 다음과 같다.

    # 제목

    ## 저자의 문제 인식 및 주장 (15문장 이내)

    ## 저자 소개
    '''

    print("문서 분석을 시작합니다...")

    response = client.models.generate_content(
        model="gemini-2.5-flash", # 현재 안정적인 모델명으로 예시 (사용자님의 2.5-flash-lite도 가능)
        contents=txt,
        config=types.GenerateContentConfig(
            temperature=0.1,
            system_instruction=system_instruction
        )
    )

    return response.text

if __name__ == '__main__':
    # 코랩에 업로드한 파일명으로 수정 (예: /content/data.txt)
    # 폴더가 없다면 미리 생성하거나 경로를 단순화하세요.
    file_path = '/content/암흑물질 탐색을 위한 저온에서 CsI 결정의 특성 연구 (1)_with_preprocessing.txt'

    summary = summarize_txt(file_path)
    print(summary)

    # 결과 저장 (현재 경로에 바로 저장)
    with open('summary_result.txt', 'w', encoding='utf-8') as f:
        f.write(summary)
```
