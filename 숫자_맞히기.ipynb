```python
#upper =>문자열 전체를 대문자로 변환
#strip => 양쪽 공백 제거
userInput = ""

while userInput != "STOP":
  userInput = input("무언가 입력하거나 입력을 멈추려면 STOP을 입력하세요.").upper().strip()
```

```python
animals = []

userInput = " "

while userInput != "":
  userInput = input("동물을 입력(끝내려면 입력 x).").upper().strip()

  if len(userInput) > 0:
    animals.append(userInput)

print(animals)
```

```python
import random

randNum = random.randrange(1,101)

userInput = ""
count = 0
while userInput != randNum:
  userInput = input()
  if userInput.isnumeric():
    userInput = int(userInput)
    if userInput < 0 or userInput >100:
      print("범위 벗어남")

    else:
      count += 1

      if userInput > randNum:
        print("down")
        print(f"시도 횟수={count}")

      elif userInput < randNum:
        print("up")
        print(f"시도 횟수={count}")

      else:
        print("정답")
        print(f"시도 횟수={count}")

  else:
    continue
```
