```python
#암호화

asciMin = 32
asciMax = 126

key = 314159
key = str(key)

message = input("암호화 할 메세지: ")

messEncr = ""

for index in range(0, len(message)):
  char = ord(message[index])
  if char < asciMin or char > asciMax:
    messEncr += message[index]
  else:
    ascNum = char + int(key[index % len(key)])

    if ascNum > asciMax:
      ascNum -= (asciMax - asciMin)
    messEncr = messEncr + chr(ascNum)
print(messEncr)
```

```python
#복호화

asciMin = 32
asciMax = 126

key = 314159
key = str(key)

message = input("복호화 할 메세지: ")

messEncr = ""

for index in range(0, len(message)):
  char = ord(message[index])
  if char < asciMin or char > asciMax:
    messEncr += message[index]
  else:
    ascNum = char - int(key[index % len(key)])

    if ascNum < asciMin:
      ascNum += (asciMax - asciMin)
    messEncr = messEncr + chr(ascNum)
print(messEncr)
```
