```python
def op(num1, num2):
  더하기 = num1 + num2
  빼기 = num1 - num2
  곱하기 = num1 * num2
  나누기 = num1 / num2
  return 더하기, 빼기, 곱하기, 나누기

a, b =map(int, input('입력:').split())
p, m, mul, d = op(a, b)
print(f'{a} + {b} = {p}')
print(f'{a} - {b} = {m}')
print(f'{a} * {b} = {mul}')
print(f'{a} / {b} = {d}')
```
